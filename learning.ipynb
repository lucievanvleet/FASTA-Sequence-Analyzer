```python
#Random Number Generation
import random
print(random.randint(1, 10)) # prints a random integer between 1 and 10
print(random.random()) # prints a random float between 0 and 1
print(random.uniform(1, 10)) # prints a random float between 1 and 10
```

```python
#DICE ROLLING SIMULATION

def fair_dice():
    return random.randint(1, 6)
print(fair_dice())

#DRAW RANDOM CARD SIMULATION
def draw_random_card():
    suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
    ranks = ["2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King", "Ace"]
    suit = random.choice(suits)
    rank = random.choice(ranks)
    return f"{rank} of {suit}"
print(draw_random_card())
```

```python
#SANGER SEQUENCING SIMULATION

def sanger_synthesis_strand(sequence):
    # Define the possible nucleotides
    nucleotides = ["A", "C", "G", "T"]
    ddNTPs = ["ddA", "ddC", "ddG", "ddT"]
    
    # Create a list to hold the synthesized strand
    synthesized_strand = []
    
    # Iterate through each nucleotide in the input sequence
    for nucleotide in sequence:
        # Randomly choose a nucleotide from the possible nucleotides or ddNTPs
        chosen_nucleotide = random.choice([nucleotides, ddNTPs])
        # Append the chosen nucleotide to the synthesized strand
        synthesized_strand.append(chosen_nucleotide)
        if chosen_nucleotide in ddNTPs:
            # If a ddNTP is chosen, terminate the synthesis
            break
    
    # Join the list into a string and return it
    return ''.join(synthesized_strand)

#run the simulation with a sample sequence
sample_sequence = "ACGTACGTACGT"
synthesized_strand = sanger_synthesis_strand(sample_sequence)
print(f"Synthesized strand: {synthesized_strand}")
```

```python
#madlib game
def madlib_game():
    print("Welcome to the Madlib Game!")
    adjective = input("Enter an adjective: ")
    noun = input("Enter a noun: ")
    verb = input("Enter a verb: ")
    adverb = input("Enter an adverb: ")
    
    madlib = f"Once upon a time, there was a {adjective} {noun} who loved to {verb} {adverb}."
    return madlib
print(madlib_game())
```
